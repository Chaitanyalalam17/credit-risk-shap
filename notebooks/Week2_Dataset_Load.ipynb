```python
from ucimlrepo import fetch_ucirepo
import pandas as pd
import numpy as np

dataset = fetch_ucirepo(id=144)
X = dataset.data.features
y = dataset.data.targets

print("UCI German Credit dataset loaded!")
print(f"Features shape: {X.shape}")
print(f"Columns: {list(X.columns)}")
print(f"\nClass distribution:")
print(y['class'].value_counts())
```

```python

print("=== DATASET INFO ===")
print(X.dtypes)
print(f"\nMissing values: {X.isnull().sum().sum()}")
print(f"\nFirst 3 rows:")
print(X.head(3))

print("\n=== KEY FEATURES ===")
print("""
Feature descriptions (UCI German Credit):
Attribute1:  Status of existing checking account
Attribute2:  Duration in months
Attribute3:  Credit history
Attribute4:  Purpose of loan
Attribute5:  Credit amount
Attribute6:  Savings account/bonds
Attribute7:  Employment since
Attribute8:  Instalment rate (% of disposable income)
Attribute9:  Personal status/sex
Attribute10: Other debtors/guarantors
Attribute11: Present residence since
Attribute12: Property
Attribute13: Age in years
Attribute14: Other instalment plans
Attribute15: Housing
Attribute16: Number of existing credits
Attribute17: Job
Attribute18: Number of liable people
Attribute19: Telephone
Attribute20: Foreign worker
""")
```

```python
import os

os.makedirs('../data/raw', exist_ok=True)

df = X.copy()
df['credit_risk'] = y['class'].values

# Recode: 1=good→0, 2=bad→1 (standard binary format)
df['credit_risk'] = df['credit_risk'].map({1: 0, 2: 1})

df.to_csv('../data/raw/german_credit.csv', index=False)

print(f"Saved to data/raw/german_credit.csv")
print(f"Shape: {df.shape}")
print(f"Class distribution (0=good, 1=bad):")
print(df['credit_risk'].value_counts())
```
